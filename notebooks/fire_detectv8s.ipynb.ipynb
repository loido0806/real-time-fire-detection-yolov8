```python
import zipfile
import os

# Trỏ đúng vào file dataset zip trên Drive
zip_path = '/content/drive/MyDrive/HUST_Fire_Project/FIRE-1.zip'
extract_path = '/content/'

print("📦 Đang giải nén dữ liệu từ Drive...")
with zipfile.ZipFile(zip_path, 'r') as zip_ref:
    zip_ref.extractall(extract_path)

print("✅ Đã giải nén thành công! Dữ liệu đã sẵn sàng ở /content/FIRE-1")
```

```python
from ultralytics import YOLO

# 1. Khởi tạo mô hình
model = YOLO('yolov8s.pt')

# 2. Bắt đầu quá trình huấn luyện
results = model.train(
    data='/content/FIRE-1/data.yaml',     # TRỎ VÀO DATA LOCAL VỪA GIẢI NÉN
    epochs=100,
    patience=30,
    imgsz=640,
    batch=-1,
    device=0,

    # --- CẤU HÌNH LƯU TRỮ ---
    project='/content/drive/MyDrive/HUST_Fire_Project',
    name='fire_detection_v8s',

    # --- CẤU HÌNH TĂNG CƯỜNG DỮ LIỆU BẮT LỬA NHỎ ---
    scale=0.5,
    mosaic=1.0
)

print("🎉 Đã huấn luyện xong mô hình YOLOv8s!")
```
